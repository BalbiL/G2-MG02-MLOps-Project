```python
import requests
import json
```

```python
def get_on_this_day_events(month, day):
    """
    Fetches events from the Wikimedia "On this day" API.
    
    Args:
        month (int): The month (1-12).
        day (int): The day (1-31).
        
    Returns:
        list: A list of event objects, or an empty list if an error occurs.
    """
    # Format month and day with leading zeros (e.g., 09 for September)
    month_str = f"{month:02d}"
    day_str = f"{day:02d}"
    
    # This is the public Wikimedia API endpoint
    url = f"https://api.wikimedia.org/feed/v1/wikipedia/en/onthisday/all/{month_str}/{day_str}"
    
    # It's good practice to set a custom User-Agent
    headers = {
        'User-Agent': 'MyNewsRecommender/1.0 (contact@example.com)'
    }
    
    try:
        response = requests.get(url, headers=headers)
        # Raise an exception for bad status codes (like 404 or 500)
        response.raise_for_status() 
        
        data = response.json()
        
        # 'events' is the key for historical events
        return data.get('events', [])
        
    except requests.exceptions.RequestException as e:
        print(f"Error fetching data from Wikimedia API: {e}")
        return []

# --- Example Usage for your 9/11 idea ---

month = 9
day = 11

print(f"Fetching events for {month}/{day}...")
events = get_on_this_day_events(month, day)

if events:
    print(f"Successfully fetched {len(events)} events.")
    
    # Now, let's find the specific event you mentioned
    for event in events:
        if event["year"] > 2000:
            print(event["text"])
            print(event["year"])
            print(event['pages'][0]["title"])
    
    
#     target_event = None
#     for event in events:
#         if "September 11 attacks" in event['text']:
#             target_event = event
#             break
            
#     if target_event:
#         print("\n--- Found Target Event ---")
#         print(f"Year: {target_event['year']}")
#         print(f"Text: {target_event['text']}")
        
#         # This is the key we need for Part 2: the Wikipedia page title
#         if target_event['pages']:
#             main_page = target_event['pages'][0]
#             page_title = main_page['title']
#             print(f"Wikipedia Page Title: {page_title}")
#         else:
#             print("Error: Event has no associated Wikipedia pages.")
            
#     else:
#         print("\nCould not find the 'September 11 attacks' event in the list.")
else:
    print("No events found for this date.")
```
